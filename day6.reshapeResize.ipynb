Reshape : Change shape of array<br>
M1<br>
- Array.reshape(shape,Fill_by)
- shape= (row,col)
- Fill_by "C" ===> Fill_by Row
-          "F" ===> Fill_by column

M2<br>
- np.reshape(data,shape,order(Fill_by))
- shape= (row,col)
- order (Fill_by) "C" ===> Fill_by Row
-          "F" ===> Fill_by column

- resshape(rows, columns) specify new shape if dimensions match

- np.reshape() returns a new reshaped view (or copy), it does NOT modify the original array in-place.

- If values is less than new shape we got error (but size fills zeros)
