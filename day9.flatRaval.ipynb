ravel() vs flatten()

Both are used to convert a multi-dimensional array into a 1D array.

##### ravel() --- view (changes original array)
##### flatten()---copy (make a copy and original array is unchanged)

##### Both convert a multi-dimensional array into a 1-D array, but the memory behavior is different.

```python
# 📌 Definitions
# Function	         What it does	                                Memory behavior
# ravel()	         Returns a flattened view (if possible)	        Shares memory
# flatten()	         Returns a flattened copy	                    New memory
```

```python
import numpy as np
```

```python
a=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
a
```

```python
b=a.raval()
print(a)
print(b)
```
