ravel() vs flatten()

Both are used to convert a multi-dimensional array into a 1D array.

##### ravel() --- view (changes original array)
##### flatten()---copy (make a copy and original array is unchanged)

##### Both convert a multi-dimensional array into a 1-D array, but the memory behavior is different.

```python
# 📌 Definitions
# Function	         What it does	                                Memory behavior
# ravel()	         Returns a flattened view (if possible)	        Shares memory
# flatten()	         Returns a flattened copy	                    New memory
```

```python
import numpy as np
```

```python
a=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
a
```

```python
# Function	         What it does	                                Memory behavior
# ravel()	         Returns a flattened view (if possible)	        Shares memory
b=a.ravel()
print(a)
print(b)
```

```python
b[1]=99
b
```

```python
a
```

```python
print(a[1][1])
```

```python
a[1][1]=100
a
```

```python
b
```

```python
# flatten()	         Returns a flattened copy	                    New memory
x=np.array([[11,22],[33,44],[55,66],[77,88]])
y=x.flatten()
print(x)
print(y)
```

```python
x[0][0]=100
x
```

```python
y
```

```python
y[0]=999
y
```

```python
x
```

```python
# Syntax

# arr.flatten(order='C')
# Order
# 'C' → Row-wise (default)
# 'F' → Column-wise
```

```python
x=np.array([[11,22],[33,44],[55,66],[77,88]])
y=x.flatten(order='C')              #c= row
print(x) 
print(y)
```

```python
x=np.array([[11,22],[33,44],[55,66],[77,88]])
y=x.flatten(order='F')             # F =column
print(x)
print(y)
```

```python
# Quick Notes
# ravel()
# -------
# - Converts ND → 1D
# - Returns View (usually)
# - Faster
# - Less memory

# flatten()
# ---------
# - Converts ND → 1D
# - Returns Copy
# - Slower
# - More memory


# | Feature                     | ravel()   | flatten() |
# | --------------------------- | --------- | --------- |
# | Converts to 1D              | ✅         | ✅         |
# | Returns View                | ✅ Usually | ❌         |
# | Returns Copy                | ❌         | ✅         |
# | Memory Efficient            | ✅         | ❌         |
# | Changes may affect original | ✅         | ❌         |


# One-Line Memory Trick
# ravel()   = View
# flatten() = Copy
```
