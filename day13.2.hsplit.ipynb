NumPy hsplit()
- hsplit() is used to split an array horizontally (column-wise).

Syntax
- np.hsplit(arr, indices_or_sections)

Parameters
- arr → Input array
- indices_or_sections
- Number of equal parts
- OR list of split indices

Rules
- ✅ Splits columns.
- ✅ Works only with 2D or higher-dimensional arrays.
- ✅ Number of columns must be divisible by the number of sections (for equal split).
- ✅ Returns a list of arrays.
- ✅ Original array is not modified.

```python
a=np.array([[1,2],[3,4],[5,6],[7,8]])
b=np.hsplit(a,2)
b
```

```python
print(a.shape)
print(b[0].shape)
print(b[0])
print(b[1])
```

```python
# Split at Specific Columns
a = np.array([[1,2,3,4],
              [5,6,7,8]])
```

```python
b=np.hsplit(a,2)
b
```

```python
#specific split
b=np.hsplit(a,[1,3])
b
```
