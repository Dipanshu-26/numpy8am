NumPy split()
- split() is used to divide an array into multiple sub-arrays.

Syntax
- np.split(arr, indices_or_sections, axis=0)

Parameters
- arr → Input array
- indices_or_sections
- Number of equal parts
- OR list of split indices

axis
- 0 → Split rows (default)
- 1 → Split columns

```python
import numpy as np
```

```python
a=np.array([1,2,3,4,5,6,7,8])
a
```

###### 1D Array
###### Split into Equal Parts

```python
b=np.split(a,2)
b
```

```python
b=np.split(a,3)
b
```

```python
b=np.split(a,4)
b
```

```python
b[1]
```

```python
# Split at Specific Index
#          0  1 2 3 4 5 6 7
a=np.array([1,2,3,4,5,6,7,8])

b=np.split(a,[2,4])
b
```

```python
# Split at Specific Index
#          0  1 2 3 4 5 6 7
a=np.array([1,2,3,4,5,6,7,8])

b=np.split(a,[2,4,7])
b
```

```python
# 2D Array
a = np.array([[1,2],[3,4],[5,6],[7,8]])
# Split Rows (axis=0)
b=np.split(a,2,axis=0)
b
```

```python
# 2D Array
a = np.array([[1,2],[3,4],[5,6],[7,8]])
# Split Rows (axis=0)
b=np.split(a,3,axis=0)
b
```

```python
# 2D Array
a = np.array([[1,2],[3,4],[5,6],[7,8]])
# Split Rows (axis=0)
b=np.split(a,4,axis=0)
b
```

```python
# 2D Array , specific split
a = np.array([[1,2],[3,4],[5,6],[7,8]])
# Split Rows (axis=0)
b=np.split(a,[1,3],axis=0)
b
```

```python
# 2D Array
a = np.array([[1,2],[3,4],[5,6],[7,8]])
# Split columns (axis=1)
b=np.split(a,2,axis=1)
b
```

```python
# 2D Array
a = np.array([[1,2,4],[3,4,4],[5,6,4],[7,8,4]])
# Split columns (axis=1)
b=np.split(a,3,axis=1)
b
```

```python
# 2D Array
a = np.array([[1,2,4,5],[3,4,4,5],[5,6,4,5],[7,8,4,5]])
# Split columns (axis=1)
b=np.split(a,[1,3],axis=1)
b
```

```python
# 3D Array
a=np.arange(24).reshape(2,3,4)
a
```

```python
# Split Blocks (axis=0)
b=np.split(a,2,axis=0)
b
```

```python
# Split rowa (axis=1)
b=np.split(a,2,axis=1)
b
```

```python
# Split rowa (axis=1)
b=np.split(a,3,axis=1)
b
```
