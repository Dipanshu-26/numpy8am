```python
b=np.array([[1,2,3,4],[1,2,3,4]])
```

```python
x=np.array([1,2,3,4])
y=np.array([[1,2,3,4],[1,2,3,4]])
z=np.array([[[1,2],[3,4],[5,6],[7,8]]])
```
