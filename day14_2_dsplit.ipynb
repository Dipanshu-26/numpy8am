```python
import numpy as np
```

```python
# NumPy dsplit()
# - dsplit() is used to split an array depth-wise (along the 3rd axis / columns in a 3D array).

# Syntax
# - np.dsplit(arr, indices_or_sections)

# Parameters
# - arr → Input array
# - indices_or_sections
# - Number of equal parts
# - OR list of split indices

# Rules
# - ✅ Splits along axis=2.(blocks)
# - ✅ Works only with 3D or higher-dimensional arrays.
# - ✅ Size of the 3rd dimension must be divisible by the number of sections (for equal split).
# - ✅ Returns a list of arrays.
# - ✅ Original array is not modified.
```

```python
a=np.arange(24).reshape(2,3,4)
a
```

```python
b=np.dsplit(a,2)
b
```

```python
b=np.dsplit(a,[1,3])
b
```

```python
b=np.dsplit(a,4)
b
```

```python
a=np.arange(24).reshape(4,2,3)
a
```

```python
b=np.dsplit(a,[1])
b
```
