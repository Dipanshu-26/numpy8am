```python
# 2. resize() - Change Shape and Size 

# array.resize() - changes original array.
# np.resize() - does NOT modify the original array.
```

```python
# Method 1 : arr.resize(shape)
# ------------------------------
# Changes original array
# Extra positions filled with 0
a=np.array([1,2,3,4,5,6])
a.resize((3,3))
a
```

```python
a=np.array([1,2,3,4,5,6])
a.resize((4,5))
a
```

```python
# Method 2: np.resize(array, shape)
# ---------------------------------
# Does NOT change original array
# Returns a new array
# Extra positions filled by repeating existing values
```

```python
a=np.array([1,2,3,4,5,6])
b=np.resize(a,(3,3))
print(b)
print(a)
```

```python
a=np.array([1,2,3,4,5,6])
b=np.resize(a,(5,6))
print(b)
```

```python
# np.pad() is the safe and professional way to add extra values around an array.
# It is much better than resize() for padding.

# 🔹 np.pad() — Add values around array
# 📌 Syntax
# np.pad(array, pad_width, mode='constant', constant_values=0)
```

```python
a=np.array([1,2,3,4,5,6])
b=np.pad(a,(2,3))
b
```

```python
a=np.array([1,2,3,4,5,6])
b=np.pad(a,(2,3),constant_values=8)
b
```

```python
c=np.pad(a,(3,3),constant_values=7)
c
```
