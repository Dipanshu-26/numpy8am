```python
# 2D padding
a=np.array([[1,2],[3,4]])

#             r     c          (row   column)
#            b a   b  a        (before after)
b=np.pad(a,((1,2),(2,3)))
b
#Meaning: ((top,bottom), (left,right))
```

```python
# 3D padding
a=np.array([[[1,2],[3,4]],
            [[5,6],[7,8]]])
a.shape
```

```python
#            b       r     c          (block, row   column)
#            b  a  b a   b  a        (before after)
b=np.pad(a,((1,1),(1,2),(2,3)))
b
# Meaning:((depth_before,depth_after), (rows_before,rows_after), (cols_before,cols_after))
```

```python
# np.pad(array, pad_width, mode='constant', constant_values=0)
```

```python
# 1. mode='constant' (Default) : Adds the same value around the array.
# --------------------------------------------------------------------
```

```python
# 2. mode='edge' : Repeats edge values
# ------------------------------------
a=np.array([11,2,3,4,5,66])
b=np.pad(a,(2,3),mode='edge')
b
```

```python
# 3. mode='wrap' : # Wraps values from the opposite side.
# ------------------------------------------------------
a=np.array([11,2,3,4,5,66])
b=np.pad(a,(2,3),mode='wrap')
b
```

```python
# 4. mode='reflect' : Reflects values without repeating edge. (mirrer efect)
# ---------------------------------------------------------------------------
a=np.array([11,2,3,4,5,66])
b=np.pad(a,(2,3),mode='reflect')
b
```

```python
# 5. mode='symmetric' : Reflects values including edge.
# --------------------------------------------------------
a=np.array([11,2,3,4,5,66])
b=np.pad(a,(2,3),mode='symmetric')
b
```

```python
# Important Note
# ---------------
# np.pad(arr,pad_width,mode='constant',constant_values=value)

# pad_width → How many values to add.
# mode → How to fill added values.
# constant_values → Used only with mode='constant'.

# Most Common Interview Modes
# | Mode        | Behavior                  |
# | ----------- | ------------------------- |
# | `constant`  | Fill with fixed value     |
# | `edge`      | Repeat edge values        |
# | `wrap`      | Repeat from opposite side |
# | `reflect`   | Mirror without edge       |
# | `symmetric` | Mirror with edge          |


# For beginners, remember: constant, edge, reflect, wrap are the four modes most commonly used in practice.
```
