```python
# 📌 Handling Missing Values in NumPy
# What are Missing Values? : Missing values are unknown, empty, or invalid data points that appear in real-world datasets.

# Examples:
# Student marks missing
# Sensor data failure
# Survey unanswered questions
# Corrupted files

# Special Missing Values in NumPy
# NumPy uses special floating-point values to represent missing/invalid data.
```

```python
# Special Value	                   Meaning
# np.nan	                       Not a Number (missing value)
# np.inf	                       Positive infinity
# -np.inf                          Negative infinity

# 🔍 Why handling is required
# Reason	                    What happens if ignored
# Math becomes wrong	        mean, sum become NaN
# ML models fail	            Cannot train
# Graphs break	                Empty charts
# Decisions become wrong	    Wrong business result
```

```python
# 🔹 Detecting Missing Values
# Function	              Use
# np.isnan()	          Find NaN
# np.isinf()	          Find infinity
# np.isfinite()	          Valid numbers

# 🔹 Handling Techniques
# Method	                      Use
# Remove missing	              np.nan_to_num()
# Replace with 0	              Fill default
# Replace with mean	              Statistical filling
# Mask invalid	                  Ignore invalid
# Interpolation	                  Estimate values
```

```python
import numpy as np
```

```python
# find nan

a = np.array([1,2,3,np.nan,4,5,np.nan])

print(np.isnan(a))
```

```python
# can you compare 2 NaN values like (np.nan == np.nan)
# answer is no we cant comapre only need to use isnan()
```

```python
#replace 
#np.nan_to_num(array, nan = value) default = 0
b = np.nan_to_num(a)
print(b)
print(a)
```

```python
b=np.nan_to_num(a,nan=100)
b
```

```python
a= np.array([1,2,3,np.inf,4,5,-np.inf,6,7])

print(np.isinf(a))
```

```python
np.isposinf(a)
```

```python
np.isneginf(a)
```

```python
b=np.nan_to_num(a,posinf=100,neginf=-100)
b
```
