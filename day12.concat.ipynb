NumPy concatenate()<br>

concatenate() is used to join two or more arrays.<br>

Syntax<br>
- np.concatenate((arr1, arr2), axis=0)<br>

Parameters
- arr1, arr2, ... → Arrays to join
- axis=0 → Join rows (default)
- axis=1 → Join columns

```python
# 1D Array
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

c=np.concat((a,b))
c
```

```python
print(a)
print(b)
print(c)
```

```python
# 2D Array
a = np.array([[1,2],
              [3,4]])

b = np.array([[5,6],
              [7,8]])
```

```python
# Concatenate Rows (axis=0)
c=np.concat((a,b),axis=0)
c
```

```python
# Concatenate Columns (axis=1)
c=np.concat((a,b),axis=1)
c
```

```python
# 3D Array
a = np.array([
    [[1,2],[3,4]]
])

b = np.array([
    [[5,6],[7,8]]
])

print(a.shape)
# (1,2,2)
```

```python
# Concatenate Along Depth
# as a block
c=np.concat((a,b),axis=0)
c
```

```python
# as a row
c=np.concat((a,b),axis=1)
c
```

```python
# as a column
c=np.concat((a,b),axis=2)
c
```
