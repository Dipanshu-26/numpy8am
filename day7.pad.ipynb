```python
# np.pad() is the safe and professional way to add extra values around an array.
# It is much better than resize() for padding.

# 🔹 np.pad() — Add values around array
# 📌 Syntax
# np.pad(array, pad_width, mode='constant', constant_values=0)
```

```python
import numpy as np
```

```python
a=np.array([1,2,3,4])
```

```python
b=np.pad(a,(2,3))
b
```

```python
b.size
```

```python
np.pad(a,(2,3),constant_values=9)
```

```python
np.pad(a,5)
```

```python
x=np.array([[1,2],[3,4]])
x
```

```python
np.pad(x,2)     #add 2 rows befor and after and 2 columns befor and after in each row
```

```python
np.pad(x,2,constant_values=8)
```

```python
#         b a
np.pad(x,(2,3))
```

```python
#          b a   b  a 
#           r     c
np.pad(x,((3,1),(4,3)))
```

```python
np.pad(x,((3,1),(4,3)),constant_values=7)
```
